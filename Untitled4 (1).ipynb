```python
import pandas as pd
import numpy as np

df = pd.read_csv("/content/titanic/train.csv")

print("Dataset shape:", df.shape)
display(df.head())
```

```python
print("Missing values in each column:")
print(df.isnull().sum())
```

```python
# Create a copy of the original dataset
df_clean = df.copy()

# Fill missing Age values with median
df_clean["Age"] = df_clean["Age"].fillna(
    df_clean["Age"].median()
)

# Fill missing Embarked values with the most frequent value
df_clean["Embarked"] = df_clean["Embarked"].fillna(
    df_clean["Embarked"].mode()[0]
)

# Drop Cabin because most of its values are missing
df_clean = df_clean.drop(columns=["Cabin"])

# Check missing values after cleaning
print("Missing values after cleaning:")
print(df_clean.isnull().sum())
```

```python
# Check duplicate rows
print("Duplicate rows:", df_clean.duplicated().sum())
```

```python
# Check data types and dataset information
df_clean.info()
```

```python
# Convert columns to appropriate data types
df_clean["PassengerId"] = df_clean["PassengerId"].astype("int64")
df_clean["Survived"] = df_clean["Survived"].astype("int64")

df_clean["Pclass"] = df_clean["Pclass"].astype("category")
df_clean["Sex"] = df_clean["Sex"].astype("category")
df_clean["Embarked"] = df_clean["Embarked"].astype("category")

# Check updated data types
print(df_clean.dtypes)
```

```python
print("Dataset shape:", df_clean.shape)

print("Total missing values:",
      df_clean.isnull().sum().sum())

print("Total duplicate rows:",
      df_clean.duplicated().sum())

display(df_clean.head())
```

```python
df_clean.to_csv("titanic_cleaned.csv", index=False)

print("Cleaned dataset saved successfully!")
```

```python

import pandas as pd

df = pd.read_csv("/content/titanic/train.csv")

print("Dataset Loaded Successfully!")
print("Dataset Shape:", df.shape)
```

```python

# Create a copy of the dataset
df_clean = df.copy()

# Fill missing Age values with median
df_clean["Age"] = df_clean["Age"].fillna(df_clean["Age"].median())

# Fill missing Embarked values with mode
df_clean["Embarked"] = df_clean["Embarked"].fillna(df_clean["Embarked"].mode()[0])

# Drop Cabin column because it has many missing values
df_clean = df_clean.drop(columns=["Cabin"])

print("Missing Values After Cleaning:")
print(df_clean.isnull().sum())

print("\nFinal Dataset Shape:", df_clean.shape)
```

```python

# Statistical Analysis

print("1. Statistical Summary")
display(df_clean.describe())

print("\n2. Survival Count")
print(df_clean["Survived"].value_counts())

print("\n3. Gender Distribution")
print(df_clean["Sex"].value_counts())

print("\n4. Passenger Class Distribution")
print(df_clean["Pclass"].value_counts())

print("\n5. Average Age")
print(df_clean["Age"].mean())

print("\n6. Average Ticket Fare")
print(df_clean["Fare"].mean())
```

```python

import matplotlib.pyplot as plt
import seaborn as sns

sns.countplot(data=df_clean, x="Survived")

plt.title("Survival Count of Titanic Passengers")
plt.xlabel("Survived (0 = No, 1 = Yes)")
plt.ylabel("Number of Passengers")

plt.show()
```

```python

sns.barplot(data=df_clean, x="Sex", y="Survived")

plt.title("Survival Rate by Gender")
plt.xlabel("Gender")
plt.ylabel("Survival Rate")

plt.show()
```

```python

sns.barplot(data=df_clean, x="Pclass", y="Survived")

plt.title("Survival Rate by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Survival Rate")

plt.show()
```

```python

plt.figure(figsize=(8, 5))

sns.histplot(data=df_clean, x="Age", bins=30, kde=True)

plt.title("Age Distribution of Titanic Passengers")
plt.xlabel("Age")
plt.ylabel("Number of Passengers")

plt.show()
```

```python

plt.figure(figsize=(8, 5))

sns.boxplot(data=df_clean, x="Survived", y="Fare")

plt.title("Fare Distribution by Survival")
plt.xlabel("Survived (0 = No, 1 = Yes)")
plt.ylabel("Ticket Fare")

plt.show()
```

## Key Insights from Exploratory Data Analysis

1. **Overall Survival:** Out of 891 passengers, 342 survived and 549 did not survive. The number of non-survivors was higher.

2. **Gender and Survival:** Female passengers had a much higher survival rate (approximately 74%) compared to male passengers (approximately 19%).

3. **Passenger Class:** First-class passengers had the highest survival rate (approximately 63%), followed by second class (47%) and third class (24%).

4. **Age Distribution:** Most passengers were young adults, particularly between 20 and 35 years of age.

5. **Fare and Survival:** Passengers who paid higher ticket fares generally had a higher median fare among survivors than non-survivors.

### Conclusion

The EDA suggests that gender, passenger class, age, and ticket fare may be useful features for building a Titanic survival prediction model.
