```python
import pandas as pd
import numpy as np

df = pd.read_csv("/content/titanic/train.csv")

print("Dataset shape:", df.shape)
display(df.head())
```

```python
print("Missing values in each column:")
print(df.isnull().sum())
```

```python
# Create a copy of the original dataset
df_clean = df.copy()

# Fill missing Age values with median
df_clean["Age"] = df_clean["Age"].fillna(
    df_clean["Age"].median()
)

# Fill missing Embarked values with the most frequent value
df_clean["Embarked"] = df_clean["Embarked"].fillna(
    df_clean["Embarked"].mode()[0]
)

# Drop Cabin because most of its values are missing
df_clean = df_clean.drop(columns=["Cabin"])

# Check missing values after cleaning
print("Missing values after cleaning:")
print(df_clean.isnull().sum())
```

```python
# Check duplicate rows
print("Duplicate rows:", df_clean.duplicated().sum())
```

```python
# Check data types and dataset information
df_clean.info()
```

```python
# Convert columns to appropriate data types
df_clean["PassengerId"] = df_clean["PassengerId"].astype("int64")
df_clean["Survived"] = df_clean["Survived"].astype("int64")

df_clean["Pclass"] = df_clean["Pclass"].astype("category")
df_clean["Sex"] = df_clean["Sex"].astype("category")
df_clean["Embarked"] = df_clean["Embarked"].astype("category")

# Check updated data types
print(df_clean.dtypes)
```

```python
print("Dataset shape:", df_clean.shape)

print("Total missing values:",
      df_clean.isnull().sum().sum())

print("Total duplicate rows:",
      df_clean.duplicated().sum())

display(df_clean.head())
```

```python
df_clean.to_csv("titanic_cleaned.csv", index=False)

print("Cleaned dataset saved successfully!")
```
